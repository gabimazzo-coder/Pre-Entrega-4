```python
import os
import json
import requests
from dotenv import load_dotenv

# Cargar variables de entorno desde el archivo .env
load_dotenv()

# Obtener la URL base desde el entorno (o usar la de JSONPlaceholder por defecto)
BASE_URL = os.getenv("BASE_URL", "https://typicode.com")
API_KEY = os.getenv("API_KEY", "") # Útil si migras a OpenWeather o similar

def extraer_datos(endpoint):
    """Realiza la petición GET a la API y maneja errores HTTP."""
    url = f"{BASE_URL}/{endpoint}"
    print(f"Iniciando extracción desde: {url}...")

    try:
        # Si la API requiriera token, se pasaría en los headers o params:
        # headers = {"Authorization": f"Bearer {API_KEY}"}
        response = requests.get(url, timeout=10)

        # Lanza una excepción si el estatus es un error (4xx o 5xx)
        response.raise_for_status()

        return response.json()

    except requests.exceptions.HTTPError as http_err:
        print(f"Error HTTP ocurrido: {http_err}")
    except requests.exceptions.ConnectionError:
        print("Error de conexión. Verifica tu red.")
    except requests.exceptions.Timeout:
        print("La petición ha superado el tiempo de espera.")
    except requests.exceptions.RequestException as err:
        print(f"Ocurrió un error inesperado: {err}")
    return None

def transformar_datos(lista_usuarios):
    """Limpia los datos extrayendo solo los campos necesarios."""
    if not lista_usuarios:
        return []

    datos_limpios = []
    print("Transformando y limpiando datos...")

    # Conservamos solo id, name, username, email y la ciudad
    for usuario in lista_usuarios:
        usuario_filtrado = {
            "id": usuario.get("id"),
            "nombre": usuario.get("name"),
            "usuario": usuario.get("username"),
            "email": usuario.get("email"),
            "ciudad": usuario.get("address", {}).get("city")
        }
        datos_limpios.append(usuario_filtrado)

    return datos_limpios

def guardar_datos(datos, nombre_archivo="data_extracted.json"):
    """Guarda la estructura de datos limpia en un archivo JSON local."""
    try:
        with open(nombre_archivo, "w", encoding="utf-8") as f:
            json.dump(datos, f, ensure_ascii=False, indent=4)
        print(f"¡Datos guardados exitosamente en '{nombre_archivo}'!")
    except IOError as e:
        print(f"Error al intentar escribir el archivo: {e}")

if __name__ == "__main__":
    # Usamos el endpoint '/users' de JSONPlaceholder como ejemplo
    datos_crudos = extraer_datos("users")

    if datos_crudos:
        datos_procesados = transformar_datos(datos_crudos)
        guardar_datos(datos_procesados)
    else:
        print("No se pudo completar el proceso debido a un error en la extracción.")
```
